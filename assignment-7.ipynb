```python
import math

def minimax(position, depth, is_maximizing_player):
    if depth == 0 or is_terminal(position):
        return evaluate(position)
        
    if is_maximizing_player:
        max_eval = -math.inf
        for child in get_children(position):
            evaluation = minimax(child, depth - 1, False)
            max_eval = max(max_eval, evaluation)
        return max_eval
    else:
        min_eval = math.inf
        for child in get_children(position):
            evaluation = minimax(child, depth - 1, True)
            min_eval = min(min_eval, evaluation)
        return min_eval

def is_terminal(position):
    return isinstance(position, int)

def evaluate(position):
    return position

def get_children(position):
    tree = {
        'A': ['B', 'C'],
        'B': [6, 7],     # Fixed: Added the missing list here
        'C': [1, 4]      
    }
    return tree.get(position, [])

best_score = minimax('A', 2, True)
print(best_score)
```
