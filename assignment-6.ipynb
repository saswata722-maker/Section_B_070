```python
import matplotlib.pyplot as plt
import networkx as nx
G = nx.DiGraph()
edges = [
    ("A", "B", "is-a", 0.9),
    ("B", "C", "has", 0.4),
    ("C", "A", "likes", 0.7),
]

for src, dst, rel, num in edges:
    G.add_edge(src, dst, label=f"{rel}\n({num})", weight=num)


pos = nx.spring_layout(G, seed=42)
nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1500,
    node_color="lightblue",
    font_weight="bold",
)
edge_labels = nx.get_edge_attributes(G, "label")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9)
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx
G = nx.DiGraph()
ordered_nodes = [
    
    ("Animal", "red"),
    ("Mammal", "blue"),
    ("Bird", "green"),
    
    ("Robin", "yellow"),
    ("Dog", "orange"),
    ("Cat", "violet"),
    
    ("Bark", "cyan"),
    ("Meow", "brown"),
    
    ("Fur", "black"),
    ("Feathers", "magenta"),
]


for node, color in ordered_nodes:
    G.add_node(node, color=color)


edges_data = [
    ("Mammal", "Animal", "is-a"),
    ("Bird", "Animal", "is-a"),
    ("Robin", "Bird", "is-a"),
    ("Dog", "Mammal", "is-a"),
    ("Cat", "Mammal", "is-a"),
    ("Bird", "Feathers", "has"),
    ("Dog", "Bark", "makes"),
    ("Cat", "Meow", "makes"),
    ("Cat", "Fur", "has"),
    ("Dog", "Fur", "has"),
    ("Bird", "Bark", "cannot"),
]


for src, dst, rel in edges_data:
    G.add_edge(src, dst, label=rel)


node_colors = [G.nodes[node]["color"] for node in G.nodes]


pos = nx.circular_layout(G)


nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    node_color=node_colors,
    font_color="white",
    font_weight="bold",
    font_size=9,
)


edge_labels = nx.get_edge_attributes(G, "label")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10)


plt.show()
```
