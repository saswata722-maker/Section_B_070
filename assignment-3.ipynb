**IMPORT OF IRIS AND TEST ACCURACY**

```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

path = "/kaggle/input/datasets/organizations/uciml/iris/Iris.csv"
iris = pd.read_csv(path)

print("Shape of the dataset:",iris.shape)
print("\nFirst  5 rows:")
print(iris.head())

x = iris.iloc[:,1:5]
y = iris["Species"]

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.2,random_state=42)

knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(x_train,y_train)
y_pred= knn.predict(x_test)
print("\n Predict Species")
print(y_pred)

accuracy = accuracy_score(y_test,y_pred)
print("\nAccuracy:",accuracy*100,"%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test,y_pred))
print("\nClassification report :")
print(classification_report(y_test,y_pred))
```

**WRITE A PYTHON PROGRAM TO USE DFS AND BFS SEARCH**

```python
import matplotlib.pyplot as plt
import networkx as nx
from collections import deque
from IPython.display import HTML
from matplotlib.animation import FuncAnimation

G = nx.Graph()
edge = {
    ("A", "B"),
    ("A", "C"),
    ("B", "D"),
    ("B", "E"),
    ("C", "F"),
    ("C", "G"),
    ("E", "H"),
}
G.add_edges_from(edge)


visited = []
parent = {}
queue = deque(["A"])

# 3. BFS Traversal Loop
while queue:
    node = queue.popleft()
    if node not in visited:
        visited.append(node)
        for neighbour in sorted(G.neighbors(node)):
            if neighbour not in visited:
                parent[neighbour] = node
                queue.append(neighbour)
pos = nx.spring_layout(G, seed=5)
fig, ax = plt.subplots(figsize=(10, 8))


# 5. Animation frame function
def update(frame):
    ax.clear()
    visited_now = visited[: frame + 1]

    
    visited_colors = [
        "red" if n in visited_now else "lightgray" for n in G.nodes()
    ]

    # Track edge coloring along the traversal tree paths
    edge_colors = []
    for u, v in G.edges():
        if (v in parent and parent[v] == u and v in visited_now) or (
            u in parent and parent[u] == v and u in visited_now
        ):
            edge_colors.append("red")
        else:
            edge_colors.append("lightgray")


    nx.draw(
        G,
        pos,
        ax=ax,
        with_labels=True,
        node_color=visited_colors,
        edge_color=edge_colors,
        node_size=1000,
        width=3,
        edgecolors="black",
    )
    ax.set_title(f"Current Node: {visited[frame]}", fontsize=14)
ani = FuncAnimation(
    fig, update, frames=len(visited), interval=1200, repeat=False
)
plt.close()
HTML(ani.to_jshtml())
```
